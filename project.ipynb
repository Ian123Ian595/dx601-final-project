### Problem 6 (20 points)

Which input column gives the best linear model of the target on its own?
How does that model compare to the model in problem 5?


```python
# YOUR CODE HERE
X = df.drop(columns='target')
y = df['target']
best_col = None
best_mse = np.inf
mse_per_col = {}
for col in X.columns:
    X_single = X[[col]]
    model_single = sklearn.linear_model.LinearRegression()
    model_single.fit(X_single, y)
    y_pred_single = model_single.predict(X_single)
    mse_single = np.mean((y-y_pred_single)**2)
    mse_per_col[col] = mse_single
    if mse_single < best_mse:
        best_mse = mse_single
        best_col = col
best_col, best_mse

# Alcohol is the input column that gives the best linear model of the target on its own.  However, the MSE for alcohol is noticeably higher than the MSE of the full multivariate model from problem 5.
```

### Problem 8 (40 points)

Perform principal components analysis of the input columns.
Compute how much of the data variation is explained by the first half of the principal components.
Build a linear regression using coordinates computed from the first half of the principal components.
Compare the mean squared error of this model to the previous model.
Plot actual targets vs predictions again. 

This problem depends on material from week 13.

```python
# YOUR CODE HERE
X = df.drop(columns='target')
y = df['target']
pca = PCA()
X_pca = pca.fit_transform(X)
n_components = X.shape[1]
k = int(math.ceil(n_components/2))
print("Number of principle components:", n_components)
print("Using first", k, "principal components")
explained_ratio = pca.explained_variance_ratio_
var_first_half = explained_ratio[:k].sum()
print(f"Variance explained by first {k} PCs: {var_first_half:3f}")
X_pca_half = X_pca[:, :k]
model_pca = sklearn.linear_model.LinearRegression()
model_pca.fit(X_pca_half, y)
y_pred_pca = model_pca.predict(X_pca_half)
mse_pca = np.mean((y-y_pred_pca)**2)
print(f"MSE using first {k} PCs: {mse_pca:.3f}")
plt.figure(figsize=(10,5))
plt.scatter(y, y_pred_pca, alpha=0.5)
plt.xlabel("Predicted Target")
plt.title(f"Actual vs Predicted Quality Using First {k} Principle Components")
min_val= min(y.min(), y_pred_pca.min())
max_val= min(y.max(), y_pred_pca.max())
plt.plot([min_val, max_val], [min_val, max_val], linestyle = '--')
plt.show()

#MSE is close to the full model's MSE, this suggest that most of the predictive information is contained in the leading components and that dimensionality reduction could be done with limited loss of accuracy.  Any increase in MSE compared to the full model represents the error introduced by discarding the lower-variance components.
```

### Problem 9 (20 points)

What pair of input columns has the highest correlation?
How is that correlation reflected in the principal components?

```python
# YOUR CODE HERE
X = df.drop(columns='target')
corr_matrix = X.corr()
abs_corr = corr_matrix.abs()
np.fill_diagonal(abs_corr.values, 0)
max_corr = abs_corr.max().max()
pair = np.where(abs_corr == max_corr)
row_idx = pair[0][0]
col_idx = pair[1][0]
col1 = abs_corr.index[row_idx]
col2 = abs_corr.columns[col_idx]
max_corr, col1, col2

loadings = pca.components_
feature_names = X.columns
for pc_idx in range(min(5, loadings.shape[0])):
    print(f"PC{pc_idx+1}: {feature_names.get_loc(col1)}={loadings[pc_idx, feature_names.get_loc(col1)]:3f},"
          f"{feature_names.get_loc(col2)} = {loadings[pc_idx, feature_names.get_loc(col2)]:.3f}")

# Density and residual sugar input columns has the highest correlation (0.84).  Indicating that these two variables tend to move together strongly.  
# In the PCA results, this strong correlation is reflected in the loading patterns of the principal components.  For example, in the first principle component, both density and residual sugar have large loadings with the same sign,
# meaning that both variables increase or decrease together.  
```

### Problem 10 (30 points)

Identify an outlier row in the data set.
You may use any criteria discussed in this module, and you must explain the criteria and how it led to picking this row.
Give a visualization showing how much this row sticks out compared to the other data based on your criteria.

```python
# YOUR CODE HERE
X = df.drop(columns='target')
z_scores = scipy.stats.zscore(X)
z_scores_abs = np.abs(z_scores)
max_z_per_row = z_scores_abs.max(axis=1)
outlier_idx = np.argmax(max_z_per_row)
outlier_row = df.iloc[outlier_idx]
outlier_idx, max_z_per_row[outlier_idx], outlier_row

feat_x = 'residual sugar'
feat_y = 'density'
plt.figure(figsize= (10,5))
plt.scatter(df[feat_x], df[feat_y], alpha= 0.4, label='All wines')
plt.scatter(outlier_row[feat_x], outlier_row[feat_y], s=100, marker='x', linewidths= 2, label='Outlier')
plt.xlabel(feat_x)
plt.ylabel(feat_y)
plt.title("Outlier in Residual Sugar vs Density")
plt.legend()
plt.show()

#I used a z-score based criterion to identify the outlier row in the dataset.  I first computed the z-score for each feature, rescaling them to have mean 0 and standard deviation 1, and then calculated each row's maximum absolute z-score to measure how extreme it was in any dimension.  
#Using this criterion, the row with index 2781 stood out with a max |z| of approximately 15.03, far larger than the rest of the dataset.  This row contains unusually extreme values, most notably
#Residual sugar (65.8), density (1.04), and volatile acidity (0.965), all of which are well outside typical ranges for white wine.
#In the scatter plot of residual sugar vs. density, this sample appears isolated fromt he main cluster of wines, visually confirming its status as an outlier under the z-score method.  
```
